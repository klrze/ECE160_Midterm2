Matrix Swap

```python
def Swap2(matrix: list, pos1: tuple, pos2: tuple) -> list:

    # extract the row and column positions from pos1 and pos2

    x1, y1 = pos1
    x2, y2 = pos2
    # print(x1, y1, x2, y2)

    # swap
    [matrix[x1][y1], matrix[x2][y2]] = [matrix[x2][y2], matrix[x1][y1]]

    return matrix


if __name__ == "__main__":

    matrix=[[1, 2, 3],[4, 5, 6],[7, 8, 9]]
    r = Swap2(matrix, (0,0), (2,2))
    print(matrix)

    # [[9, 2, 3],[4, 5, 6],[7, 8, 1]]

    # your function should work with the following example as well
    matrix=[['a', 'b', 'c'],['d', 'e', 'f'],['g', 'h', 'i']]
    r = Swap2(matrix, (1,2), (2,0))
    print(matrix)
```

Matrix Multiplication

```python
A = [
    [1, 2, 3],
    [4, 5, 6]
]  # 2x3

B = [
    [7, 8],
    [9, 10],
    [11, 12]
]  # 3x2

rows_A = len(A)
cols_A = len(A[0])
rows_B = len(B)
cols_B = len(B[0])

# 1. Verify inner dimensions match
if cols_A != rows_B:
    print("Cannot multiply: incompatible dimensions")
else:
    # 2. Initialize result matrix (rows_A x cols_B) filled with 0s
    result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]

    # 3. Triple loop: row of A -> col of B -> shared inner index k
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j] += A[i][k] * B[k][j]

    for row in result:
        print(" ".join([str(x) for x in row]))
```

```python
def MultiplyMatrices(matrix1, matrix2):
    
    rows1 = len(matrix1)
    cols1 = len(matrix1[0])
    rows2 = len(matrix2)
    cols2 = len(matrix2[0])

    # verify if matrix multiplication can be done by checking dimensions
    if cols1 != rows2:
        return 'Error: Not Possible'
    else:
        # resultant matrix (rows1 x cols2) with '0' placeholders
        result = [[0 for i in range(cols2)] for x in range(rows1)]

        for i in range(rows1):
            for j in range(cols2):
                for k in range(cols1):
                    result[i][j] += matrix1[i][k] * matrix2[k][j]

        return result

    

if __name__ == '__main__':

    print(MultiplyMatrices([[1, 2],[3, 4]], [[5, 6],[7, 8]]))

    print(MultiplyMatrices([[1, 2],[3, 4], [5, 6]], [[7, 7, 9],[10, 11, 12]]))

    print(MultiplyMatrices([[1, 2],[3, 4], [5, 6]], [[7, 7, 9],[10, 11, 12], [13, 14, 15]]))
```

Transposing a matrix

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

rows = len(matrix)     # 2
cols = len(matrix[0])  # 3

transposed = []
for c in range(cols):          # Outer loop: c = 0, 1, 2
    new_row = []
    for r in range(rows):      # Inner loop: r = 0, 1
        new_row.append(matrix[r][c])
    transposed.append(new_row)

for row in transposed:
    print(" ".join([str(x) for x in row]))
```

```python
class RightTriangle:
    
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def __str__(self):
        return f'Base: {self.base}, Height: {self.height}'

    def Area(self):
        return 1/2 * self.base * self.height

    def Perimeter(self):
        c = (self.base ** 2 + self.height ** 2) ** 0.5
        return self.base + self.height + c
    
    def __add__(self, other):
        if isinstance(other, RightTriangle):
            return self.Perimeter() + other.Perimeter()
        elif type(other) == float:
            return self.Area() + other
        elif type(other) == int:
            return self.Perimeter() + other


if __name__ == "__main__":
    # 1. Test __init__
    t1 = RightTriangle(3, 4)    # Area = 6.0, Perimeter = 12.0 (3 + 4 + 5)
    t2 = RightTriangle(5, 12)   # Area = 30.0, Perimeter = 30.0 (5 + 12 + 13)

    # 2. Test __str__
    print("--- Testing __str__ ---")
    print("t1:", t1)
    print("t2:", t2)

    # 3. Test Area() and Perimeter()
    print("\n--- Testing Area & Perimeter ---")
    print("t1 Area (expected 6.0):", t1.Area())
    print("t1 Perimeter (expected 12.0):", t1.Perimeter())
    print("t2 Area (expected 30.0):", t2.Area())
    print("t2 Perimeter (expected 30.0):", t2.Perimeter())

    # 4. Test __add__ operator overrides
    print("\n--- Testing + Operator ---")
    print("RightTriangle + RightTriangle (expected 42.0):", t1 + t2)   # 12.0 + 30.0
    print("RightTriangle + float (expected 10.5):", t1 + 4.5)          # 6.0 + 4.5
    print("RightTriangle + int (expected 22.0):", t1 + 10)             # 12.0 + 10
```

```python
class Circle:
    def __init__(self, pos, r):
        self.pos = pos
        self.x, self.y = pos
        self.r = r

    def Area(self):
        return 3.141592653589793 * (self.r ** 2)

    def isIntersecting(self, other):
        # Distance between the two centers: sqrt((x2 - x1)^2 + (y2 - y1)^2)
        dx = other.x - self.x
        dy = other.y - self.y
        distance = (dx ** 2 + dy ** 2) ** 0.5

        # Circles intersect/touch if distance between centers <= sum of radii
        return distance <= (self.r + other.r)

    def __str__(self):
        return f"Circle at ({self.x}, {self.y}) with radius {self.r}"


if __name__ == "__main__":
    # Create 3 circles
    c1 = Circle((0, 0), 5)
    c2 = Circle((6, 0), 3)    # Distance to c1 is 6; sum of radii is 8 -> Intersects
    c3 = Circle((20, 20), 2)  # Far away -> Does NOT intersect

    # 1. Test __str__
    print("--- Testing __str__ ---")
    print("c1:", c1)
    print("c2:", c2)
    print("c3:", c3)

    # 2. Test Area()
    print("\n--- Testing Area() ---")
    print(f"c1 Area (expected ~78.54): {c1.Area():.2f}")
    print(f"c2 Area (expected ~28.27): {c2.Area():.2f}")
    print(f"c3 Area (expected ~12.57): {c3.Area():.2f}")

    # 3. Test isIntersecting()
    print("\n--- Testing isIntersecting() ---")
    print("c1 and c2 intersecting (expected True):", c1.isIntersecting(c2))
    print("c1 and c3 intersecting (expected False):", c1.isIntersecting(c3))
    print("c2 and c3 intersecting (expected False):", c2.isIntersecting(c3))
```
